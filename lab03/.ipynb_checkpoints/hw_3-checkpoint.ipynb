## 1 Задача 1.3.12
Знайти ймовiрнiсть того, що серед трьох цифр, кожна з яких вибрана навмання, буде лише 1, 2, 3 рiзних.

Обчисліть теоретичну (повним перебором) та еміричну (симулюванням $100000$ експериментів) імовірності.

```python
import numpy as np
from itertools import product


def print_arr(arr):
    for i in range(len(arr)):
        print(f"{i+1}: {prob[i]}", end="; ")
    print()


#  Theoretical
digits = [i for i in range(10)]
prob = np.zeros(3)
combinations = list(product(digits, repeat=3))
combinations_len = len(combinations)
for c in combinations:
    prob[len(set(c)) - 1] += 1
prob /= combinations_len
print("Theoretical probability:", end=" ")
print_arr(prob)

#  Empirical
np.random.seed(0)
samples = np.random.randint(0, 10, size=(100000, 3))
empirical = np.zeros(3)
unique_counts = np.array([len(set(row)) for row in samples])
for i in range(1, 4):
    empirical[i - 1] = np.sum(unique_counts == i)
empirical /= 100000

result = np.vstack([theoretical, empirical])
print("Empiric probability:", end=" ")
print_arr(empirical)
```

# 2 Задача 1.3.14
З послiдовностi чисел $1, 2, . . . , n$ вибирають навмання $k$ рiзних чисел.
Яка ймовiрнiсть того, що:
1. кожне з вибраних чисел кратне даному числу $p$;
2. кожне з вибраних чисел кратне хоча б одному з двох взаємно простих чисел $p$ i $q$;
3. серед вибраних чисел є хоча б одне кратне $p$?

Напишіть  відповідні функції для обрахунку теоретичної (повним перебором) та еміричної (симулюванням $100000$ експериментів) імовірностей в залежності від параметрів $n, k, p, q$.
Виведіть результат для
- $n = 25, k = 5, p = 3, q = 4$;
- $n = 25, k = 10, p = 3, q = 4$.

```python
import numpy as np
from itertools import combinations

def theoretical_probabilities(n, k, p, q):
    all_combinations = list(combinations(range(1, n+1), k))
    total = len(all_combinations)
    count1 = 0
    count2 = 0
    count3 = 0
    for combo in all_combinations:
        mods = np.array(combo)
        if np.all(mods % p == 0):
            count1 += 1
        if np.all((mods % p == 0) | (mods % q == 0)):
            count2 += 1
        if np.any(mods % p == 0):
            count3 += 1
    return np.array([count1, count2, count3]) / total

def empirical_probabilities(n, k, p, q, trials=100000, seed=0):
    np.random.seed(seed)
    successes = np.zeros(3)
    for _ in range(trials):
        sample = np.random.choice(np.arange(1, n+1), size=k, replace=False)
        if np.all(sample % p == 0):
            successes[0] += 1
        if np.all((sample % p == 0) | (sample % q == 0)):
            successes[1] += 1
        if np.any(sample % p == 0):
            successes[2] += 1
    return successes / trials

params = [
    (25, 5, 3, 4),
    (25, 10, 3, 4)
]

for n, k, p, q in params:
    theo = theoretical_probabilities(n, k, p, q)
    emp = empirical_probabilities(n, k, p, q)
    print(f"\nn={n}, k={k}, p={p}, q={q}")
    for i, desc in enumerate([
        "1. Усі вибрані числа кратні p",
        "2. Усі вибрані числа кратні хоча б одному з p або q",
        "3. Серед вибраних є хоча б одне число, кратне p"
    ]):
        print(f"{desc}: Теоретична = {theo[i]:.8f}, Емпірична = {emp[i]:.8f}")
```

# 3 Задача 1.4.6
В коморi знаходяться $n$ пар черевикiв.
З них випадковим чином вибираються $2k$ черевикiв.
Яка ймовiрнiсть того, що серед вибраних черевикiв:
1. вiдсутнi парнi;
2. є рiвно одна комплектна пара;
3. є двi комплектнi пари?

Напишіть  відповідні функції для обрахунку теоретичної (повним перебором) та еміричної (симулюванням $100000$ експериментів) імовірностей в залежності від параметрів $n, k$.
Виведіть результат для
- $n = 8, k = 4$;
- $n = 13, k = 5$.

```python
import numpy as np
from itertools import combinations

def theoretical_probabilities(n, k):
    shoes = []
    for i in range(n):
        shoes.append((i, 'L'))
        shoes.append((i, 'R'))
    all_combinations = list(combinations(shoes, 2 * k))
    total = len(all_combinations)
    count0 = 0
    count1 = 0
    count2 = 0
    for combo in all_combinations:
        ids = {}
        for pair in combo:
            key = pair[0]
            ids[key] = ids.get(key, 0) + 1
        full_pairs = sum(1 for v in ids.values() if v == 2)
        if full_pairs == 0:
            count0 += 1
        if full_pairs == 1:
            count1 += 1
        if full_pairs == 2:
            count2 += 1
    return np.array([count0, count1, count2]) / total

def empirical_probabilities(n, k, trials=100000, seed=0):
    np.random.seed(seed)
    shoes = []
    for i in range(n):
        shoes.append((i, 'L'))
        shoes.append((i, 'R'))
    successes = np.zeros(3)
    for _ in range(trials):
        sample = np.random.choice(len(shoes), size=2 * k, replace=False)
        selected = [shoes[i] for i in sample]
        ids = {}
        for pair in selected:
            key = pair[0]
            ids[key] = ids.get(key, 0) + 1
        full_pairs = sum(1 for v in ids.values() if v == 2)
        if full_pairs == 0:
            successes[0] += 1
        if full_pairs == 1:
            successes[1] += 1
        if full_pairs == 2:
            successes[2] += 1
    return successes / trials

params = [
    (8, 4),
    (13, 5)
]

for n, k in params:
    theo = theoretical_probabilities(n, k)
    emp = empirical_probabilities(n, k)
    print(f"\nn={n}, k={k}")
    for i, desc in enumerate([
        "1. Відсутні парні пари",
        "2. Є рівно одна комплектна пара",
        "3. Є рівно дві комплектні пари"
    ]):
        print(f"{desc}: Теоретична = {theo[i]:.5f}, Емпірична = {emp[i]:.5f}")
```

# 4 Задача 1.2.15
Нехай $\Omega = {1, 2, \ldots, 2n}$.
Всiм числам приписанi ймовiрностi, пропорцiйнi логарифмам цих чисел.
Знайти цi ймовiрностi.
Знайти ймовiрнiсть того, що в результатi експерименту з’явиться:
1. парне число;
2. непарне число.

Напишіть функцію для обрахунку еміричної (симулюванням $100000$ експериментів) імовірності в залежності від параметра $n$.
Виведіть результат для
- $n = 10$;
- $n = 25$.

```python
import numpy as np

def compute_probabilities(n):
    omega = np.arange(1, 2 * n + 1)
    logs = np.log(omega)
    probs = logs / logs.sum()
    return omega, probs

def empirical_parity_probability(n, trials=100000, seed=0):
    omega, probs = compute_probabilities(n)
    np.random.seed(seed)
    samples = np.random.choice(omega, size=trials, p=probs)
    even = np.sum(samples % 2 == 0) / trials
    odd = np.sum(samples % 2 == 1) / trials
    return even, odd

params = [10, 25]

for n in params:
    omega, probs = compute_probabilities(n)
    even, odd = empirical_parity_probability(n)
    print(f"\nn={n}")
    print("Ймовірності, пропорційні логарифмам:")
    for i in range(len(omega)):
        print(f"{omega[i]}: {probs[i]:.5f}", end="; ")
    print(f"\n1. Ймовірність парного числа (емпірична): {even:.5f}")
    print(f"2. Ймовірність непарного числа (емпірична): {odd:.5f}")
```
