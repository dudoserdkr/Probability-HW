## 1 Від'ємний біноміальний розподіл

Більш детальна інформація в [документації](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.nbinom.html).

```python
from scipy.stats import nbinom
```

```python
n = 5
p = 0.4
```

### 1.1 Щільність

```python
x = np.arange(-0.5, n*5+1, 0.5)
```

Виведіть значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
for point in x:
    print(f"t={point}, pmf(t, n, p)={nbinom.pmf(point, n, p)}")
```

Зобразіть графічно значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
def plot(x, y, width=6, height=2):
    plt.figure(figsize=(width, height))
    plt.stem(x, y)
    plt.grid(True)
    plt.show()

pmf_values_x = nbinom.pmf(x, n, p)
plot(x, pmf_values_x, 10, 3)
```

Зобразіть графічно значення функції щільності від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x1$.

```python
x1 = np.arange(0, n*5+1)
```

```python
pmf_values_x1 = nbinom.pmf(x1, n, p)
plot(x1, pmf_values_x1, 10, 4)
```

### 1.2 Функція розподілу

Виведіть значення функції розподілу від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
for point in x:
    print(f"x={point}, F(x)={nbinom.cdf(point, n, p)}")
```

Зобразіть графічно значення функції розподілу від'ємного біноміального розподілу з параметрами $n, p$ в точках вектора $x$.

```python
cdf_values_x = nbinom.cdf(x, n, p)

plot(x, cdf_values_x, 11, 5)
```

### 1.3 Випадкові величини

Згенеруйте $100$ випадкових величин з від'ємного біноміального розподілу з параметрами $n, p$.

```python
random_values = nbinom.rvs(n, p, size=100)

print(random_values)
print()
```

Виведіть кількість кожного елемента в отриманому випадковому векторі.

```python
each_rv_count = {}
for rv in random_values:
    each_rv_count[rv] = each_rv_count.get(rv, 0) + 1

for key, val in each_rv_count.items():
    print(f"Value={key}, count={val}")
```

Зобразіть графічно кількість кожного елемента в отриманому випадковому векторі.

```python
plot(each_rv_count.keys(), each_rv_count.values(), 15, 3)
```

## 3 Різні параметри

### 3.1 Біноміальний розподіл

Зобразіть графічно функцію щільності, функцію розподілу та гістограму $10000$ випадкових величин біноміального розподілу з параметрами:
1. $n = 15, p = 0.8$
2. $n = 15, p = 0.2$
3. $n = 20, p = 0.4$
4. $n = 20, p = 0.6$
5. $n = 7, p = 0.5$
6. $n = 25, p = 0.5$
7. $n = 100, p = 0.5$

```python
from scipy.stats import binom


def plot(x, y, title, width=6, height=2):
    plt.figure(figsize=(width, height))
    plt.stem(x, y)
    plt.grid(True)
    plt.title(title)
    plt.show()

def plot_discrete_hist(data, title, density=True):
    min_val = int(np.min(data))
    max_val = int(np.max(data))
    bins = np.arange(min_val, max_val + 2)

    plt.figure(figsize=(10, 6))
    plt.hist(data, bins=bins, density=density, align='left', edgecolor='black', rwidth=0.8)
    plt.xticks(bins[:-1])
    plt.title(title)
    plt.grid(True)
    plt.show()

input_values = [(15, 0.8), (15, 0.2), (20, 0.4), (20, 0.6), (7, 0.5), (25, 0.5), (100, 0.5)]

for (n,p) in input_values:
    x = np.arange(0, n + 1)
    y_pmf = binom.pmf(x, n, p)
    plot(x, y_pmf, f"Binom(n={n}, p={p}) PMF", width=8, height=3)

    y_cdf = binom.cdf(x, n, p)

    plot(x, y_cdf, f"Binom(n={n}, p={p}) CDF", 8, 3)

    rvs = binom.rvs(n, p, size=10000)
    plot_discrete_hist(rvs, f"Histogram of 10000 random values for Binom(n={n}, p={p})")
```

### 3.2 Геометричний розподіл

Зобразіть графічно функцію щільності, функцію розподілу та гістограму $10000$ випадкових величин геометричного розподілу з параметрами:
1. $p = 0.1$
2. $p = 0.2$
3. $p = 0.3$
4. $p = 0.4$
5. $p = 0.5$
6. $p = 0.6$
7. $p = 0.7$
8. $p = 0.8$
9. $p = 0.9$

```python
from scipy.stats import geom

input_values = [0.1 * i for i in range(1, 10)]

n = 50
for p in input_values:
    x = np.arange(1, n)
    y_pmf = geom.pmf(x, p)
    plot(x, y_pmf, f"Geom(p={p}) PMF", width=8, height=3)

    y_cdf = geom.cdf(x, p)

    plot(x, y_cdf, f"Geom(p={p}) CDF", 8, 3)

    rvs = binom.rvs(n, p, size=10000)
    plot_discrete_hist(rvs, f"Histogram of 10000 random values for Geom(p={p})")
```

## 4 Апроксимація біноміального розподілу розподілом Пуассона

Покажіть схожість біноміального розподілу та розподілу Пуассона (їх функцій щільності) з відповідними параметрами.
Наприклад:
- $n = 100, p = 0.05$
- $n = 100, p = 0.1$
- $n = 100, p = 0.15$
- $n = 100, p = 0.2$
- $n = 100, p = 0.35$
- $n = 100, p = 0.5$
- $n = 100, p = 0.75$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

input_values = [
    (100, 0.05),
    (100, 0.1),
    (100, 0.15),
    (100, 0.2),
    (100, 0.35),
    (100, 0.5),
    (100, 0.75)
]

for n, p in input_values:
    lam = n * p
    k = np.arange(0, n + 1)
    pmf_binom = binom.pmf(k, n, p)
    pmf_poisson = poisson.pmf(k, lam)
    plt.figure(figsize=(8, 4))
    plt.plot(k, pmf_binom, 'o-', label=f'Біном. (n={n}, p={p})')
    plt.plot(k, pmf_poisson, 's--', label=f'Пуассон (λ={lam:.2f})')
    plt.title('Порівняння функцій щільності: Біноміальний vs Пуассонівський розподіл')
    plt.xlabel('k')
    plt.ylabel('P(X = k)')
    plt.grid(True)
    plt.legend()
    plt.show()
```

$\textbf{Висновок.}$ При великих значеннях $n$ та малих $p$, біноміальний розподіл з параметрами $n$ та $p$ добре апроксимується пуассонівським розподілом з параметром $\lambda = n\cdot p$. На графіках видно, що функції щільності для обох розподілів збігаються при $p \leq 1$, що підтверджує використання пуассонівської апроксимації для розрахунку біноміальних ймовірностей за умов великих $n$.
